```python
import pandas as pd
import numpy as np

np.random.seed(42)

dates = pd.date_range(
    start="2026-01-01",
    periods=90,
    freq="D"
)

traffic = np.random.normal(18000, 1200, 90).astype(int)

conversion_rate = np.random.normal(0.022, 0.002, 90)

orders = (traffic * conversion_rate).astype(int)

revenue_per_order = np.random.normal(2500, 150, 90)

revenue = (orders * revenue_per_order).astype(int)

marketing_cost = np.random.normal(40000, 3000, 90).astype(int)

refunds = np.random.normal(20, 5, 90).astype(int)

df = pd.DataFrame({
    "date": dates,
    "traffic": traffic,
    "orders": orders,
    "revenue": revenue,
    "marketing_cost": marketing_cost,
    "refunds": refunds
})

df.head()
```

```python
df.loc[40, "traffic"] = 40000
```

```python
df["Conversion rate"] = (df["orders"]/df["traffic"] *100)
```
